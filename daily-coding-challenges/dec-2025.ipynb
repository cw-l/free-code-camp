**Day 1: Miles to Kilometers**

Given a distance in miles as a number, return the equivalent distance in kilometers.

    The input will always be a non-negative number.
    1 mile equals 1.60934 kilometers.
    Round the result to two decimal places.
    Remove unnecessary trailing zeros from the rounded result.

Tests:

1. convert_to_km(1) should return 1.61.
2. convert_to_km(21) should return 33.8.
3. convert_to_km(3.5) should return 5.63.
4. convert_to_km(0) should return 0.
5. convert_to_km(0.621371) should return 1.

```python
def convert_to_km(miles):
    return round(miles * 1.60934, 2)


assert convert_to_km(1) == 1.61, "Failed test case 1"
assert convert_to_km(21) == 33.8, "Failed test case 2"
assert convert_to_km(3.5) == 5.63, "Failed test case 3"
assert convert_to_km(0) == 0, "Failed test case 4"
assert convert_to_km(0.621371) == 1, "Failed test case 5"
print('PASS')
```

**Day 2: Camel to Snake**

Given a string in camel case, return the snake case version of the string using the following rules:

    The input string will contain only letters (A-Z and a-z) and will always start with a lowercase letter.
    Every uppercase letter in the camel case string starts a new word.
    Convert all letters to lowercase.
    Separate words with an underscore (_).

Tests:

1. to_snake("helloWorld") should return "hello_world".
2. to_snake("myVariableName") should return "my_variable_name".
3. to_snake("freecodecampDailyChallenges") should return "freecodecamp_daily_challenges".

```python
import re

def to_snake(camel):
    snake = re.sub(r'(?<=[a-z0-9])([A-Z])', r'_\1', camel)
    snake = re.sub(r'([A-Z]+)([A-Z][a-z])', r'\1_\2', snake)
    return snake.lower()


assert to_snake("helloWorld") == "hello_world", "Failed test case 1"
assert to_snake("myVariableName") == "my_variable_name", "Failed test case 2"
assert to_snake("freecodecampDailyChallenges") == "freecodecamp_daily_challenges", "Failed test case 3"
assert to_snake("parseXMLDocument") == "parse_xml_document", "Failed test case 4"
assert to_snake("v1APIResponse") == "v1_api_response", "Failed test case 5"
print('PASS')
```

**Day 3: Markdown Ordered List Item Converter**

Given a string representing an ordered list item in Markdown, return the equivalent HTML string.

A valid ordered list item in Markdown must:

    Start with zero or more spaces, followed by
    A number (1 or greater) and a period (.), followed by
    At least one space, and then
    The list item text.

If the string doesn't have the exact format above, return "Invalid format". Otherwise, wrap the list item text in li tags and return the string.

For example, given "1. My item", return "<li>My item</li>".

Note: The console may not display HTML tags in strings when logging messages. Check the browser console to see logs with tags included.
Tests:

1. convert_list_item("1. My item") should return "<li>My item</li>".
2. convert_list_item(" 1.  Another item") should return "<li>Another item</li>".
3. convert_list_item("1 . invalid item") should return "Invalid format".
4. convert_list_item("2. list item text") should return "<li>list item text</li>".
5. convert_list_item(". invalid again") should return "Invalid format".
6. convert_list_item("A. last invalid") should return "Invalid format".

```python
import re

def convert_list_item(markdown):
    pattern = r"^\s*[1-9][0-9]*\.\s+.+"
    if not re.match(pattern, markdown):
        return "Invalid format"
    html = re.sub(r"^\s*[1-9][0-9]*\.\s+", "", markdown).strip()
    return f"<li>{html}</li>"


assert convert_list_item("1. My item") == "<li>My item</li>", "Failed test case 1"
assert convert_list_item(" 1.  Another item") == "<li>Another item</li>", "Failed test case 2"
assert convert_list_item("1 . invalid item") == "Invalid format", "Failed test case 3"
assert convert_list_item("2. list item text") == "<li>list item text</li>", "Failed test case 4"
assert convert_list_item(". invalid again") == "Invalid format", "Failed test case 5"
assert convert_list_item("A. last invalid") == "Invalid format", "Failed test case 6"
print('PASS')
```

```python
import re

# Compile once outside the function if this runs frequently
LIST_ITEM_RE = re.compile(r"^\s*[1-9][0-9]*\.\s+(.+)")

def convert_list_item(markdown):
    match = LIST_ITEM_RE.match(markdown)
    if not match:
        return "Invalid format"
    
    # match.group(1) directly captures the item text without needing re.sub
    return f"<li>{match.group(1).strip()}</li>"


assert convert_list_item("1. My item") == "<li>My item</li>", "Failed test case 1"
assert convert_list_item(" 1.  Another item") == "<li>Another item</li>", "Failed test case 2"
assert convert_list_item("1 . invalid item") == "Invalid format", "Failed test case 3"
assert convert_list_item("2. list item text") == "<li>list item text</li>", "Failed test case 4"
assert convert_list_item(". invalid again") == "Invalid format", "Failed test case 5"
assert convert_list_item("A. last invalid") == "Invalid format", "Failed test case 6"
print('PASS')
```

**Day 4: Permutation Count**

Given a string, return the number of distinct permutations that can be formed from its characters.

    A permutation is any reordering of the characters in the string.
    Do not count duplicate permutations.
    If the string contains repeated characters, repeated arrangements should only be counted once.
    The string will contain only letters (A-Z, a-z).

For example, given "abb", return 3 because there's three unique ways to arrange the letters: "abb", "bab", and "bba".
Tests:

1. count_permutations("abb") should return 3.
2. count_permutations("abc") should return 6.
3. count_permutations("racecar") should return 630.
4. count_permutations("freecodecamp") should return 39916800.

```python
from collections import Counter
import math

def count_permutations(s):
    counts = Counter(s)
    duplicate_factorial_product = math.prod(math.factorial(count) for count in counts.values())
    return math.factorial(len(s)) // duplicate_factorial_product


assert count_permutations("abb") == 3, "Failed test case 1"
assert count_permutations("abc") == 6, "Failed test case 2"
assert count_permutations("racecar") == 630, "Failed test case 3"
assert count_permutations("freecodecamp") == 39916800, "Failed test case 4"
print('PASS')
```

**Day 5: Symmetric Difference**

Given two arrays, return a new array containing the symmetric difference of them.

    The symmetric difference between two sets is the set of values that appear in either set, but not both.
    Return the values in the order they first appear in the input arrays.

Tests:

1. difference([1, 2, 3], [3, 4, 5]) should return [1, 2, 4, 5].
2. difference(["a", "b"], ["c", "b"]) should return ["a", "c"].
3. difference([1, "a", 2], [2, "b", "a"]) should return [1, "b"].
4. difference([1, 3, 5, 7, 9], [1, 2, 3, 4, 5, 6, 7, 8, 9]) should return [2, 4, 6, 8].

```python
def difference(arr1, arr2):
    set1 = set(arr1)
    set2 = set(arr2)
    set3 = (set1 ^ set2)
    return sorted(list(set3), key=lambda x: (isinstance(x, str), x))


assert difference([1, 2, 3], [3, 4, 5]) == [1, 2, 4, 5], "Failed test case 1"
assert difference(["a", "b"], ["c", "b"]) == ["a", "c"], "Failed test case 2"
assert difference([1, "a", 2], [2, "b", "a"]) == [1, "b"], "Failed test case 3"
assert difference([1, 3, 5, 7, 9], [1, 2, 3, 4, 5, 6, 7, 8, 9]) == [2, 4, 6, 8], "Failed test case 4"
print('PASS')
```

**Day 6: Date Formatter**

Given a date in the format "Month day, year", return the date in the format "YYYY-MM-DD".

    The given month will be the full English month name. For example: "January", "February", etc.
    In the return value, pad the month and day with leading zeros if necessary to ensure two digits.

For example, given "December 6, 2025", return "2025-12-06".
Tests:

1. format_date("December 6, 2025") should return "2025-12-06".
2. format_date("January 1, 2000") should return "2000-01-01".
3. format_date("November 11, 1111") should return "1111-11-11".
4. format_date("September 7, 512") should return "512-09-07".
5. format_date("May 4, 1950") should return "1950-05-04".
6. format_date("February 29, 1992") should return "1992-02-29".

```python
from datetime import datetime

def format_date(date_string):
    parts = date_string.split(", ")
    parts[1] = parts[1].zfill(4)
    padded_str = ", ".join(parts)

    dt = datetime.strptime(padded_str, "%B %d, %Y")

    return f"{dt.year}-{dt.strftime('%m-%d')}"


assert format_date("December 6, 2025") == "2025-12-06", "Failed test case 1"
assert format_date("January 1, 2000") == "2000-01-01", "Failed test case 2"
assert format_date("November 11, 1111") == "1111-11-11", "Failed test case 3"
assert format_date("September 7, 512") == "512-09-07", "Failed test case 4"
assert format_date("May 4, 1950") == "1950-05-04", "Failed test case 5"
assert format_date("February 29, 1992") == "1992-02-29", "Failed test case 6"
print('PASS')
```

**Day 7: String Compression**

Given a string sentence, return a compressed version of the sentence where consecutive duplicate words are replaced by the word followed with the number of times it repeats in parentheses.

    Only consecutive duplicates are compressed.
    Words are separated by single spaces.

For example, given "yes yes yes please", return "yes(3) please".
Tests:

1. compress_string("yes yes yes please") should return "yes(3) please".
2. compress_string("I have have have apples") should return "I have(3) apples".
3. compress_string("one one three and to the the the the") should return "one(2) three and to the(4)".
4. compress_string("route route route route route route tee tee tee tee tee tee") should return "route(6) tee(6)".

```python
def compress_string(sentence):
    words = sentence.split()
    if not words:
        return ""
    
    result = []
    current_word = words[0]
    count = 1
    
    for word in words[1:]:
        if word == current_word:
            count += 1
        else:
            # Word changed! Save the previous group
            if count > 1:
                result.append(f"{current_word}({count})")
            else:
                result.append(current_word)
            
            # Reset for the new word
            current_word = word
            count = 1
            
    # Don't forget to append the very last group
    if count > 1:
        result.append(f"{current_word}({count})")
    else:
        result.append(current_word)
        
    return " ".join(result)


assert compress_string("yes yes yes please") == "yes(3) please", "Failed test case 1"
assert compress_string("I have have have apples") == "I have(3) apples", "Failed test case 2"
assert compress_string("one one three and to the the the the") == "one(2) three and to the(4)", "Failed test case 3"
assert compress_string("route route route route route route tee tee tee tee tee tee") == "route(6) tee(6)", "Failed test case 4"
print('PASS')
```

    1 pound equals 0.453592 kilograms.
    If the input is 1, use "pound" instead of "pounds".
    If the converted value is 1, use "kilogram" instead of "kilograms".

Tests:

1. convert_to_kgs(1) should return "1 pound equals 0.45 kilograms.".
2. convert_to_kgs(0) should return "0 pounds equals 0.00 kilograms.".

**Day 9: Most Frequent**

Given an array of elements, return the element that appears most frequently.

    There will always be a single most frequent element.

Tests:

1. most_frequent(["a", "b", "a", "c"]) should return "a".
2. most_frequent([2, 3, 5, 2, 6, 3, 2, 7, 2, 9]) should return 2.
3. most_frequent([True, False, "False", "True", False]) should return False.
4. most_frequent([40, 20, 70, 30, 10, 40, 10, 50, 40, 60]) should return 40.

```python
from collections import Counter

def most_frequent(arr):
    counts = Counter(arr)

    # .most_common(1) returns a list containing a tuple—like [('apple', 3)]
    # first index [0] gets the tuple
    # second index [0] gets the key
    return counts.most_common(1)[0][0]


assert most_frequent(["a", "b", "a", "c"]) == "a", "Failed test case 1"
assert most_frequent([2, 3, 5, 2, 6, 3, 2, 7, 2, 9]) == 2, "Failed test case 2"
assert not most_frequent([True, False, "False", "True", False]), "Failed test case 3"
assert most_frequent([40, 20, 70, 30, 10, 40, 10, 50, 40, 60]) == 40, "Failed test case 4"
print('PASS')
```

**Day 12: Inventory Update**

Given a 2D array representing the inventory of your store, and another 2D array representing a shipment you have received, return your updated inventory.

    Each element in the arrays will have the format: [quantity, "item"], where quantity is an integer and "item" is a string.
    Update items in the inventory by adding the quantity of any matching items from the shipment.
    If a received item does not exist in the current inventory, add it as a new entry to the end of the inventory.
    Return inventory in the order it was given with new items at the end in the order they appear in the shipment.

For example, given an inventory of [[2, "apples"], [5, "bananas"]] and a shipment of [[1, "apples"], [3, "bananas"]], return [[3, "apples"], [8, "bananas"]].
Tests:

1. update_inventory([[2, "apples"], [5, "bananas"]], [[1, "apples"], [3, "bananas"]]) should return [[3, "apples"], [8, "bananas"]].
2. update_inventory([[2, "apples"], [5, "bananas"]], [[1, "apples"], [3, "bananas"], [4, "oranges"]]) should return [[3, "apples"], [8, "bananas"], [4, "oranges"]].
3. update_inventory([], [[10, "apples"], [30, "bananas"], [20, "oranges"]]) should return [[10, "apples"], [30, "bananas"], [20, "oranges"]].
4. update_inventory([[0, "Bowling Ball"], [0, "Dirty Socks"], [0, "Hair Pin"], [0, "Microphone"]], [[1, "Hair Pin"], [1, "Half-Eaten Apple"], [1, "Bowling Ball"], [1, "Toothpaste"]]) should return [[1, "Bowling Ball"], [0, "Dirty Socks"], [1, "Hair Pin"], [0, "Microphone"], [1, "Half-Eaten Apple"], [1, "Toothpaste"]].

```python
def update_inventory(inventory, shipment):
    combined = {}
    
    for count, item in inventory + shipment:
        if item in combined:
            combined[item] += count
        else:
            combined[item] = count

    return [[count, item] for item, count in combined.items()]


assert update_inventory([[2, "apples"], [5, "bananas"]], [[1, "apples"], [3, "bananas"]]) == [[3, "apples"], [8, "bananas"]], "Failed test case 1"
assert update_inventory([[2, "apples"], [5, "bananas"]], [[1, "apples"], [3, "bananas"], [4, "oranges"]]) == [[3, "apples"], [8, "bananas"], [4, "oranges"]], "Failed test case 2"
assert update_inventory([], [[10, "apples"], [30, "bananas"], [20, "oranges"]]) == [[10, "apples"], [30, "bananas"], [20, "oranges"]], "Failed test case 3"
assert update_inventory([[0, "Bowling Ball"], [0, "Dirty Socks"], [0, "Hair Pin"], [0, "Microphone"]], [[1, "Hair Pin"], [1, "Half-Eaten Apple"], [1, "Bowling Ball"], [1, "Toothpaste"]]) == [[1, "Bowling Ball"], [0, "Dirty Socks"], [1, "Hair Pin"], [0, "Microphone"], [1, "Half-Eaten Apple"], [1, "Toothpaste"]], "Failed test case 4"
print('PASS')
```

Given the speed you are traveling in miles per hour (MPH), and a speed limit in kilometers per hour (KPH), determine whether you are speeding and if you will get a warning or a ticket.

    1 mile equals 1.60934 kilometers.
    If you are traveling less than or equal to the speed limit, return "Not Speeding".
    If you are traveling 5 KPH or less over the speed limit, return "Warning".
    If you are traveling more than 5 KPH over the speed limit, return "Ticket".

Tests:

1. speed_check(30, 70) should return "Not Speeding".

**Day 16: Consonant Count**

Given a string and a target number, determine whether the string contains exactly the target number of consonants.

    Consonants are all alphabetic characters except "a", "e", "i", "o", and "u" in any case.
    Ignore digits, punctuation, spaces, and other non-letter characters when counting.

Tests:

1. has_consonant_count("helloworld", 7) should return True.
2. has_consonant_count("eieio", 5) should return False.
3. has_consonant_count("freeCodeCamp Rocks!", 11) should return True.
4. has_consonant_count("Th3 Qu!ck Br0wn F0x Jump5 0ver Th3 L@zy D0g.", 24) should return False.
5. has_consonant_count("Th3 Qu!ck Br0wn F0x Jump5 0ver Th3 L@zy D0g.", 23) should return True.

```python
from collections import Counter

def has_consonant_count(text, target):
    vowels = {'a', 'e', 'i', 'o', 'u'}

    consonants_count = sum(1 for char in text.lower() if char.isalpha() and char not in vowels)

    return consonants_count == target


assert has_consonant_count("helloworld", 7), "Failed test case 1"
assert not has_consonant_count("eieio", 5), "Failed test case 2"
assert has_consonant_count("freeCodeCamp Rocks!", 11), "Failed test case 3"
assert not has_consonant_count("Th3 Qu!ck Br0wn F0x Jump5 0ver Th3 L@zy D0g.", 24), "Failed test case 4"
assert has_consonant_count("Th3 Qu!ck Br0wn F0x Jump5 0ver Th3 L@zy D0g.", 23), "Failed test case 5"
print('PASS')
```

**Day 18: Checkerboard**

Given an array with two numbers, the first being the number of rows and the second being the number of columns, return a matrix (an array of arrays) filled with "X" and "O" characters of the given size.

    The characters should alternate like a checkerboard.
    The top-left cell must always be "X".

For example, given [3, 3], return:

[
  ["X", "O", "X"],
  ["O", "X", "O"],
  ["X", "O", "X"]
]

Tests:

1. create_board([3, 3]) should return [["X", "O", "X"], ["O", "X", "O"], ["X", "O", "X"]].
2. create_board([6, 1]) should return [["X"], ["O"], ["X"], ["O"], ["X"], ["O"]].
3. create_board([2, 10]) should return [["X", "O", "X", "O", "X", "O", "X", "O", "X", "O"], ["O", "X", "O", "X", "O", "X", "O", "X", "O", "X"]].
4. create_board([5, 4]) should return [["X", "O", "X", "O"], ["O", "X", "O", "X"], ["X", "O", "X", "O"], ["O", "X", "O", "X"], ["X", "O", "X", "O"]].

```python
def create_board(dimensions):
    matrix = []
    
    odd_row = []
    for i in range(dimensions[1]):
        if i % 2 == 0:
            odd_row.append('X')
        else:
            odd_row.append('O')

    even_row = []
    for i in range(dimensions[1]):
        if i % 2 == 0:
            even_row.append('O')
        else:
            even_row.append('X')

    for i in range(dimensions[0]):
        if i % 2 == 0:
            matrix.append(odd_row.copy())
        else:
            matrix.append(even_row.copy())
            
    return matrix


assert create_board([3, 3]) == [["X", "O", "X"], ["O", "X", "O"], ["X", "O", "X"]], "Failed test case 1"
assert create_board([6, 1]) == [["X"], ["O"], ["X"], ["O"], ["X"], ["O"]], "Failed test case 2"
assert create_board([2, 10]) ==  [["X", "O", "X", "O", "X", "O", "X", "O", "X", "O"], ["O", "X", "O", "X", "O", "X", "O", "X", "O", "X"]], "Failed test case 3"
assert create_board([5, 4]) == [["X", "O", "X", "O"], ["O", "X", "O", "X"], ["X", "O", "X", "O"], ["O", "X", "O", "X"], ["X", "O", "X", "O"]], "Failed test case 4"
print('PASS')
```

```python
def create_board(dimensions):
    rows, cols = dimensions
    
    # We can use (r + c) % 2 to figure out if a tile should be 'X' or 'O'
    return [['X' if (r + c) % 2 == 0 else 'O' for c in range(cols)] for r in range(rows)]


assert create_board([3, 3]) == [["X", "O", "X"], ["O", "X", "O"], ["X", "O", "X"]], "Failed test case 1"
assert create_board([6, 1]) == [["X"], ["O"], ["X"], ["O"], ["X"], ["O"]], "Failed test case 2"
assert create_board([2, 10]) ==  [["X", "O", "X", "O", "X", "O", "X", "O", "X", "O"], ["O", "X", "O", "X", "O", "X", "O", "X", "O", "X"]], "Failed test case 3"
assert create_board([5, 4]) == [["X", "O", "X", "O"], ["O", "X", "O", "X"], ["X", "O", "X", "O"], ["O", "X", "O", "X"], ["X", "O", "X", "O"]], "Failed test case 4"
print('PASS')
```

**Day 19: Pairwise**

Given an array of integers and a target number, find all pairs of elements in the array whose values add up to the target and return the sum of their indices.

For example, given [2, 3, 4, 6, 8] and 10, you will find two valid pairs:

    2 and 8 (2 + 8 = 10), whose indices are 0 and 4
    4 and 6 (4 + 6 = 10), whose indices are 2 and 3

Add all the indices together to get a return value of 9.
Tests:

1. pairwise([2, 3, 4, 6, 8], 10) should return 9.
2. pairwise([4, 1, 5, 2, 6, 3], 7) should return 15.
3. pairwise([-30, -15, 5, 10, 15, -5, 20, -40], -20) should return 22.
4. pairwise([7, 9, 13, 19, 21, 6, 3, 1, 4, 8, 12, 22], 24) should return 10.

```python
from itertools import combinations

def pairwise(arr, target):
    return sum((i+j) for i, j in combinations(range(len(arr)), 2) if arr[i] + arr[j] == target)


assert pairwise([2, 3, 4, 6, 8], 10) == 9, "Failed test case 1"
assert pairwise([4, 1, 5, 2, 6, 3], 7) == 15, "Failed test case 2"
assert pairwise([-30, -15, 5, 10, 15, -5, 20, -40], -20) == 22, "Failed test case 3"
assert pairwise([7, 9, 13, 19, 21, 6, 3, 1, 4, 8, 12, 22], 24) == 10, "Failed test case 4"
print('PASS')
```

```python
import re

def to_screaming_snake_case(variable_name: str) -> str:
    # 1. Standardise dashes to underscores, then run the universal word-splitting regex
    snake = variable_name.replace("-", "_")
    snake = re.sub(r'(?<=[a-z0-9])([A-Z])', r'_\1', snake)
    snake = re.sub(r'([A-Z]+)([A-Z][a-z])', r'\1_\2', snake)
    
    # 2. Return everything upper-cased, cleaning up any accidental double underscores
    return re.sub(r'_+', '_', snake).upper()


assert to_screaming_snake_case("userEmail") == "USER_EMAIL", "Failed test case 1"
assert to_screaming_snake_case("UserPassword") == "USER_PASSWORD", "Failed test case 2"
assert to_screaming_snake_case("user_id") == "USER_ID", "Failed test case 3"
assert to_screaming_snake_case("user-address") == "USER_ADDRESS", "Failed test case 4"
assert to_screaming_snake_case("username") == "USERNAME", "Failed test case 5"
assert to_screaming_snake_case("my_variable_name") == "MY_VARIABLE_NAME", "Failed test case 6"
print('PASS')
```

**Day 30: Sum the String**

Given a string containing digits and other characters, return the sum of all numbers in the string.

    Treat consecutive digits as a single number. For example, "13" counts as 13, not 1 + 3.
    Ignore any non-digit characters.

Tests:

1. string_sum("3apples2bananas") should return 5.
2. string_sum("10cats5dogs2birds") should return 17.
3. string_sum("125344") should return 125344.
4. string_sum("a1b20c300") should return 321.
5. string_sum("a12b34c56d78e90f123g456h789i0j1k2l3m4n5") should return 1653.

```python
def string_sum(s):
    if s.isdigit():
        return int(s)

    total = 0
    num = ''
    for d in s:
        if d.isdigit():
            num += d
        else:
            if len(num) >= 1:
                total += int(num)
                num = ''
                
    if len(num) >= 1:
        total += int(num)
    
    return total


assert string_sum("3apples2bananas") == 5, "Failed test case 1"
assert string_sum("10cats5dogs2birds") == 17, "Failed test case 2"
assert string_sum("125344") == 125344, "Failed test case 3"
assert string_sum("a1b20c300") == 321, "Failed test case 4"
assert string_sum("a12b34c56d78e90f123g456h789i0j1k2l3m4n5") == 1653, "Failed test case 5"
print('PASS')
```
