**Day 1: Schema Validator Part 1**

Given an object (JavaScript) or dictionary (Python), determine if it matches the following schema:

{
  username: string
}

    Extra keys are allowed

Tests:

1. is_valid_schema({"username": "bob"}) should return True.
2. is_valid_schema({"username": "jen", "posts": 30}) should return True.
3. is_valid_schema({"username": ""}) should return True.
4. is_valid_schema({"username": 7}) should return False.
5. is_valid_schema({"posts": 25}) should return False.

```python
def is_valid_schema(obj):
    if type(obj) is not dict:
        return False
        
    if type(obj.get("username")) is not str:
        return False

    return True


assert is_valid_schema({"username": "bob"}), "Failed test case 1"
assert is_valid_schema({"username": "jen", "posts": 30}), "Failed test case 2"
assert is_valid_schema({"username": ""}), "Failed test case 3"
assert not is_valid_schema({"username": 7}), "Failed test case 4"
assert not is_valid_schema({"posts": 25}), "Failed test case 5"
print('PASS') 
```

```python
import jsonschema
from jsonschema import validate

def is_valid_schema(obj):
    user_schema = {
        "type": "object",
        "properties": {
            "username": {"type": "string"},
        },
        "required": ["username"]
    }
    try:
        validate(instance=obj, schema=user_schema)
        return True
    except jsonschema.exceptions.ValidationError:
        return False


assert is_valid_schema({"username": "bob"}), "Failed test case 1"
assert is_valid_schema({"username": "jen", "posts": 30}), "Failed test case 2"
assert is_valid_schema({"username": ""}), "Failed test case 3"
assert not is_valid_schema({"username": 7}), "Failed test case 4"
assert not is_valid_schema({"posts": 25}), "Failed test case 5"
print('PASS')
```

**Day 7: Last Load**

Given the number of scoops of laundry detergent you have remaining and an array of how many scoops you used in each of the previous days, return the number of full days of detergent you have remaining.

Calculate your average daily usage from the usage history and assume that amount of usage each day going forward.

Tests:

1. last_load_date(10, [2, 2, 2, 2, 2, 2, 2]) should return 5.
2. last_load_date(16, [2, 3, 0, 3, 4, 2, 1]) should return 7.
3. last_load_date(33, [5, 0, 4, 3, 3, 2]) should return 11.
4. last_load_date(50, [2, 0, 2, 9, 12, 0, 2]) should return 12.

```python
def last_load_date(scoops, usage):
    return scoops // (sum(usage) / len(usage))


assert last_load_date(10, [2, 2, 2, 2, 2, 2, 2]) == 5, "Failed test case 1"
assert last_load_date(16, [2, 3, 0, 3, 4, 2, 1]) == 7, "Failed test case 2"
assert last_load_date(33, [5, 0, 4, 3, 3, 2]) == 11, "Failed test case 3"
assert last_load_date(50, [2, 0, 2, 9, 12, 0, 2]) == 12, "Failed test case 4"
assert last_load_date(20, [13, 9, 12, 10, 8]) == 1, "Failed test case 5"
print('PASS')
```

**Day 8: Jet Lagged**

Given a departure city, an arrival city, a flight duration in hours, and a direction of travel, return the number of jet lag hours the traveller is experiencing.

The given cities will be from the following list that includes their UTC offset:

| City | Offset |
| ---- | ------ |
| "Los Angeles" | -8 |
| "New York" | -5 |
| "London" | 0 | 
| "Istanbul" | +3 |
| "Dubai" | +4 |
| "Hong Kong" | +8 |
| "Tokyo" | +9 |

To calculate jet lag hours:

    Find the timezone difference in hours between the two cities.
    Determine the direction multiplier. If travelling "east", it's 1.5, otherwise, it's 1.0.
    Get the jet lag hours with the formula: timezone difference + (flight duration * 0.1) * direction multiplier

Return the jet lag hours rounded to one decimal place.
Tests:

1. get_jet_lag_hours("Istanbul", "Hong Kong", 10, "east") should return 6.5.
2. get_jet_lag_hours("London", "New York", 8, "west") should return 5.8.
3. get_jet_lag_hours("Hong Kong", "Tokyo", 4, "east") should return 1.6.
4. get_jet_lag_hours("Dubai", "London", 7, "west") should return 4.7.
5. get_jet_lag_hours("Los Angeles", "Hong Kong", 15, "west") should return 17.5.
6. get_jet_lag_hours("Tokyo", "Dubai", 9, "west") should return 5.9.
7. get_jet_lag_hours("New York", "Istanbul", 10, "east") should return 9.5.

```python
DIRECTION_MULT = {
    "east": 1.5, "west": 1.0
}

CITIES = {
    "los angeles": -8, "new york": -5, "london": 0, "istanbul": 3, "dubai": 4, "hong kong": 8, "tokyo": 9
}

def get_jet_lag_hours(departure_city, arrival_city, flight_duration, direction):
    return abs(CITIES.get(departure_city.lower()) - CITIES.get(arrival_city.lower())) + flight_duration * 0.1 * DIRECTION_MULT.get(direction.lower())


assert get_jet_lag_hours("Istanbul", "Hong Kong", 10, "east") == 6.5, "Failed test case 1"
assert get_jet_lag_hours("London", "New York", 8, "west") == 5.8, "Failed test case 2"
assert get_jet_lag_hours("Hong Kong", "Tokyo", 4, "east") == 1.6, "Failed test case 3"
assert get_jet_lag_hours("Dubai", "London", 7, "west") == 4.7, "Failed test case 4"
assert get_jet_lag_hours("Los Angeles", "Hong Kong", 15, "west") == 17.5, "Failed test case 5"
assert get_jet_lag_hours("Tokyo", "Dubai", 9, "west") == 5.9, "Failed test case 6"
assert get_jet_lag_hours("New York", "Istanbul", 10, "east") == 9.5, "Failed test case 7"
print('PASS')
```

**Day 13: Zoning Regulations**

Given a 2D grid (array of arrays) representing a city's building layout, return the coordinates of all buildings that are violating zoning rules.

Each cell in the grid contains one of the labels from the table below. A building is in violation if any of its (up to) 4 neighbors, horizontal or vertical, are a type it cannot be adjacent to.

| Label | Type | Cannot be adjacent to |
| ----- | ---- | --------------------- |
| "i" |	industrial | "R", "I" |
| "A" |	Agricultural | "C" |
| "R" |	Residential | "i", "C" |
| "I" |	Institutional | "i" |
| "C" |	Commercial | "R", "A" |
| "" | (empty string) | undeveloped | no restrictions |

Return the coordinates of all violating cells as an array of [row, col] pairs, in any order. If no violations exist, return an empty array.

Tests:

1. get_zone_violations([["R", "C"], ["", "C"]]) should return [[0, 0], [0, 1]].
2. get_zone_violations([["", "i"], ["", "R"], ["R", "I"]]) should return [[0, 1], [1, 1]].
3. get_zone_violations([["A", "i", "C"], ["A", "", "C"], ["R", "R", "I"]]) should return [].
4. get_zone_violations([["R", "R", "C", "R", "R"], ["R", "I", "C", "", "A"], ["R", "R", "", "i", "A"]]) should return [[0, 1], [0, 2], [0, 3]].
5. get_zone_violations([["R", "A", "A", "", "i", "i"], ["R", "I", "", "C", "i", "i"], ["R", "", "C", "C", "A", "A"], ["R", "R", "C", "I", "R", "R"]]) should return [[2, 3], [2, 4], [3, 1], [3, 2]].

```python
def get_zone_violations(grid):
    # 1. Structure the O(1) forbidden lookup map
    RULES = {
        "i": {"R", "I"},
        "A": {"C"},
        "R": {"i", "C"},
        "I": {"i"},
        "C": {"R", "A"},
        "": set()  # Undeveloped has no restrictions
    }
    
    rows = len(grid)
    cols = len(grid[0]) if rows > 0 else 0
    violations = []
    
    # 2. Sequential matrix scan
    for r in range(rows):
        for c in range(cols):
            current_label = grid[r][c]
            forbidden_neighbors = RULES.get(current_label, set())
            
            # 3. Coordinate offset stream (Up, Down, Left, Right)
            for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
                nr, nc = r + dr, c + dc
                
                # 4. Native boundary guard
                if 0 <= nr < rows and 0 <= nc < cols:
                    neighbor_label = grid[nr][nc]
                    
                    # If a violation is hit, log coordinate and break to avoid duplicates
                    if neighbor_label in forbidden_neighbors:
                        violations.append([r, c])
                        break 
                        
    return violations


assert get_zone_violations([["R", "C"], ["", "C"]]) == [[0, 0], [0, 1]], "Failed test case 1"
assert get_zone_violations([["", "i"], ["", "R"], ["R", "I"]]) == [[0, 1], [1, 1]], "Failed test case 2"
assert get_zone_violations([["A", "i", "C"], ["A", "", "C"], ["R", "R", "I"]]) == [], "Failed test case 3"
assert get_zone_violations([["R", "R", "C", "R", "R"], ["R", "I", "C", "", "A"], ["R", "R", "", "i", "A"]]) == [[0, 1], [0, 2], [0, 3]], "Failed test case 4"
assert get_zone_violations([["R", "A", "A", "", "i", "i"], ["R", "I", "", "C", "i", "i"], ["R", "", "C", "C", "A", "A"], ["R", "R", "C", "I", "R", "R"]]) == [[2, 3], [2, 4], [3, 1], [3, 2]], "Failed test case 5"
print('PASS')
```

**Day 17: Spellcaster**

Given a string of spell codes you are casting, calculate the total score.

Each character in the string represents a spell:
| Code | Spell | Category | Base Score |
| ---- | ----- | -------- | ---------- |
| "f" | Fire | Destruction | 3 |
| "l" | Lightning | Destruction | 3 |
| "i" | Ice | Control | 2 |
| "w" | Wind | Control | 2 |
| "h" |	Heal | Restoration | 1 |
| "s" | Shield | Restoration | 1 |

A combo multiplier is applied based on how many spells in a row have been cast from different categories:

    The first spell always scores at base value.
    Each consecutive spell from a different category than the previous increases the multiplier by 1.
    Casting a spell from the same category as the previous resets the multiplier back to 1.
    The score for each spell is its base score multiplied by the current multiplier.

Return the total score from the sequence of spells.

Tests:

1. cast("fihwl") should return 33.
2. cast("lwswfi") should return 45.
3. cast("wislhfl") should return 37.
4. cast("sihwlih") should return 50.

```python
BASE_SCORE = {
    "f": 3, "l": 3, "i": 2, "w": 2, "h": 1, "s": 1
}

CAT = {
    "f": "destruction", "l": "destruction",
    "i": "control", "w": "control",
    "h": "restoration", "s": "restoration" 
}

def cast(spells):
    multipler = 1
    total_score = 0

    for i in range(len(spells)):
        score = BASE_SCORE.get(spells[i])
        if i > 0 and CAT.get(spells[i]) != CAT.get(spells[i-1]):
            multipler += 1
        else:
            multipler = 1
        total_score += score * multipler

    return total_score


assert cast("fihwl") == 33, "Failed test case 1"
assert cast("lwswfi") == 45, "Failed test case 2"
assert cast("wislhfl") == 37, "Failed test case 3"
assert cast("sihwlih") == 50, "Failed test case 4"
assert cast("wishlfihwslwifihl") == 101, "Failed test case 5"
print('PASS')
```

A prime factorization is the set of prime numbers that multiply together to produce the given integer. Each number has exactly one set. For example, the prime factorization of 20 is [2, 2, 5] because 2 * 2 * 5 = 20.

If the given integer is itself prime, return it in a single-element array.

Tests:

1. prime_factorization(20) should return [2, 2, 5].
2. prime_factorization(17) should return [17].
3. prime_factorization(15) should return [3, 5].
4. prime_factorization(35) should return [5, 7].
5. prime_factorization(999) should return [3, 3, 3, 37].
6. prime_factorization(360) should return [2, 2, 2, 3, 3, 5].
7. prime_factorization(510510) should return [2, 3, 5, 7, 11, 13, 17].

```python
def prime_factorization(n):
    factors = []
    factor = 2
    
    while factor * factor <= n:
        while n % factor == 0:
            factors.append(factor)
            n //= factor
        factor += 1

    if n > 1:
        factors.append(n)
        
    return factors


assert prime_factorization(20) == [2, 2, 5], "Failed test case 1"
assert prime_factorization(17) == [17], "Failed test case 2"
assert prime_factorization(15) == [3, 5], "Failed test case 3"
assert prime_factorization(35) == [5, 7], "Failed test case 4"
assert prime_factorization(999) == [3, 3, 3, 37], "Failed test case 5"
assert prime_factorization(360) == [2, 2, 2, 3, 3, 5], "Failed test case 6"
assert prime_factorization(510510) == [2, 3, 5, 7, 11, 13, 17], "Failed test case 7"
print('PASS')
```

1. parse_frontmatter("---\ntitle: My Post\ndraft: false\nviews: 100\n---") should return { title: "My Post", draft: False, views: 100 }.
2. parse_frontmatter("---\nid: 6a174db57256a112f932195c\ntitle: My Book\nlocale: en\nwordCount: 10000\npublished: false\n---") should return { id: "6a174db57256a112f932195c", title: "My Book", locale: "en", wordCount: 10000, published: False }.
3. parse_frontmatter("---\nversion: 1.0.0\nurl: https://example.com\nprivate: true\n---") should return { version: "1.0.0", url: "https://example.com", private: True }.
4. parse_frontmatter("---\nrating: 4.5\nprice: 9.99\n---") should return { rating: 4.5, price: 9.99 }.

```python
BOOLEAN_MAP = {"true": True, "false": False}

def parse_frontmatter(s):
    fm_dict = {}
    contents = s.split("\n")
    for content in contents:
        if content != "---":
            key, value = content.split(": ", 1)
            
            value = value.strip('"').strip("'")
            
            lowercase_value = value.lower()
            if lowercase_value in BOOLEAN_MAP:
                value = BOOLEAN_MAP[lowercase_value]
            else:
                try:
                    value = int(value)
                except ValueError:
                    try:
                        value = float(value)
                    except ValueError:
                        pass
            fm_dict[key] = value
    return fm_dict


assert parse_frontmatter("---\ntitle: My Post\ndraft: false\nviews: 100\n---") == { "title": "My Post", "draft": False, "views": 100 }, "Failed test case 1"
assert parse_frontmatter("---\nid: 6a174db57256a112f932195c\ntitle: My Book\nlocale: en\nwordCount: 10000\npublished: false\n---") == { "id": "6a174db57256a112f932195c", "title": "My Book", "locale": "en", "wordCount": 10000, "published": False }, "Failed test case 2"
assert parse_frontmatter("---\nversion: 1.0.0\nurl: https://example.com\nprivate: true\n---") == { "version": "1.0.0", "url": "https://example.com", "private": True }, "Failed test case 3"
assert parse_frontmatter("---\nrating: 4.5\nprice: 9.99\n---") == { "rating": 4.5, "price": 9.99 }, "Failed test case 4"
print('PASS')
```

8. get_periodic_spelling("value") should return [].

```python
def get_periodic_spelling(word):
    periodic_table = ["H","He","Li","Be","B","C","N","O","F","Ne","Na","Mg","Al","Si","P","S","Cl","Ar","K","Ca","Sc","Ti","V","Cr","Mn","Fe","Co","Ni","Cu","Zn","Ga","Ge","As","Se","Br","Kr","Rb","Sr","Y","Zr","Nb","Mo","Tc","Ru","Rh","Pd","Ag","Cd","In","Sn","Sb","Te","I","Xe","Cs","Ba","La","Ce","Pr","Nd","Pm","Sm","Eu","Gd","Tb","Dy","Ho","Er","Tm","Yb","Lu","Hf","Ta","W","Re","Os","Ir","Pt","Au","Hg","Tl","Pb","Bi","Po","At","Rn","Fr","Ra","Ac","Th","Pa","U","Np","Pu","Am","Cm","Bk","Cf","Es","Fm","Md","No","Lr","Rf","Db","Sg","Bh","Hs","Mt","Ds","Rg","Cn","Nh","Fl","Mc","Lv","Ts","Og"]
    periodic_set = {el.lower(): el for el in periodic_table}
    
    word_lower = word.lower()
    memo = {}  # Caches index results to guarantee rapid speed

    def solve(i):
        # Base Case: Successfully reached the end of the word
        if i == len(word_lower):
            return []
        
        # Return cached result if we've been to this index before
        if i in memo:
            return memo[i]
        
        # 1. Try a 1-letter element
        one_letter = word_lower[i:i+1]
        if one_letter in periodic_set:
            rest = solve(i + 1)
            if rest is not None:
                memo[i] = [periodic_set[one_letter]] + rest
                return memo[i]
                
        # 2. Try a 2-letter element
        two_letter = word_lower[i:i+2]
        if len(two_letter) == 2 and two_letter in periodic_set:
            rest = solve(i + 2)
            if rest is not None:
                memo[i] = [periodic_set[two_letter]] + rest
                return memo[i]
        
        # No paths worked from this point forward
        memo[i] = None
        return None

    result = solve(0)
    return result if result is not None else []


assert get_periodic_spelling("neon") == ["Ne", "O", "N"], "Failed test case 1"
assert get_periodic_spelling("rational") == ["Ra", "Ti", "O", "N", "Al"], "Failed test case 2"
assert get_periodic_spelling("yarn") == ["Y", "Ar", "N"], "Failed test case 3"
assert get_periodic_spelling("carbon") in (["C", "Ar", "B", "O", "N"], ["Ca", "Rb", "O", "N"]), "Failed test case 4"
assert get_periodic_spelling("noisy") in (["N", "O", "I", "S", "Y"], ["No", "I", "S", "Y"]), "Failed test case 5"
assert get_periodic_spelling("bicycles") in (["B", "I", "C", "Y", "Cl", "Es"], ["Bi", "C", "Y", "Cl", "Es"]), "Failed test case 6"
assert get_periodic_spelling("optics") in (["O", "P", "Ti", "C", "S"], ["O", "P", "Ti", "Cs"], ["O", "Pt", "I", "C", "S"], ["O", "Pt", "I", "Cs"]), "Failed test case 7"
assert get_periodic_spelling("value") == [], "Failed test case 8"
print('PASS')
```

```python
from dataclasses import dataclass

@dataclass
class MusicProfile:
    mood: str
    genre: str
    bpm_min: int
    bpm_max: int

music_data = [
    MusicProfile("focus", "classical",  60, 109),
    MusicProfile("focus", "electronic", 60,  89),
    MusicProfile("happy", "pop",        60, 180),
    MusicProfile("happy", "classical", 110, 180),
    MusicProfile("happy", "rock",       60, 129),
    MusicProfile("happy", "electronic", 90, 134),
    MusicProfile("hype",  "rock",      130, 180),
    MusicProfile("hype",  "electronic",135, 180)
]

def get_mood(genre, bpm):
    return next((mp.mood for mp in music_data if mp.genre == genre and mp.bpm_min <= bpm <= mp.bpm_max), None)


assert get_mood("rock", 111) == "happy", "Failed test case 1"
assert get_mood("electronic", 74) == "focus", "Failed test case 2"
assert get_mood("classical", 180) == "happy", "Failed test case 3"
assert get_mood("rock", 155) == "hype", "Failed test case 4"
assert get_mood("electronic", 90) == "happy", "Failed test case 5"
assert get_mood("classical", 67) == "focus", "Failed test case 6"
assert get_mood("pop", 100) == "happy", "Failed test case 7"
assert get_mood("electronic", 135) == "hype", "Failed test case 8"
print('PASS')
```

**Day 30: Duplicate Character Count**

Given two strings, return a count of characters from the second string that can be found in the first.

    Duplicate characters in the second string are counted separately.

Tests:

1. duplicate_character_count("aloha", "hei") should return 1.
2. duplicate_character_count("jambo", "bonjour") should return 4.
3. duplicate_character_count("hello", "hola") should return 3.
4. duplicate_character_count("ola", "hej") should return 0.
5. duplicate_character_count("ciao", "konnichiwa") should return 5.
6. duplicate_character_count("merhaba", "xin chao") should return 2.
7. duplicate_character_count("hello world", "hello to everyone around the world") should return 26.

```python
def duplicate_character_count(str1, str2):
    return len([char for char in str2 if char in str1])


assert duplicate_character_count("aloha", "hei") == 1, "Failed test case 1"
assert duplicate_character_count("jambo", "bonjour") == 4, "Failed test case 2"
assert duplicate_character_count("hello", "hola") == 3, "Failed test case 3"
assert duplicate_character_count("ola", "hej") == 0, "Failed test case 4"
assert duplicate_character_count("ciao", "konnichiwa") == 5, "Failed test case 5"
assert duplicate_character_count("merhaba", "xin chao") == 2, "Failed test case 6"
assert duplicate_character_count("hello world", "hello to everyone around the world") == 26, "Failed test case 7"
print('PASS')
```
